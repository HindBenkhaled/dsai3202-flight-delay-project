## Load Silver Parquet

```python
from pyspark.sql.functions import col, trim, count, when

flights_parquet = spark.read.parquet(
    "abfss://lakehouse@airlinedata60104384.dfs.core.windows.net/processed/flights/"
)

# GUI table
display(flights_parquet)

flights_parquet.show(5, truncate=False)

# Schema + count
flights_parquet.printSchema()
print(f"Total rows: {flights_parquet.count()}")
```

```python


print("Basic Profiling\n")

# Count missing values in each column
for c in flights_parquet.columns:
    missing = flights_parquet.filter(
        col(c).isNull() |
        (col(c) == "") |
        (col(c) == "NA")
    ).count()
    print(f"{c}: {missing}")
```

## Data cleaning


### Convert delay columns to numeric & fill missing with 0

```python

df = flights_parquet

delay_cols = ["CarrierDelay", "LateAircraftDelay", "NASDelay",
              "SecurityDelay", "WeatherDelay"]

for c in delay_cols:
    df = df.withColumn(
        c,
        when(col(c).isNull(), 0)
        .when(col(c) == "", 0)
        .when(col(c) == "NA", 0)
        .otherwise(col(c).cast("double"))
    )

# Quick check
df.select("CarrierDelay", "WeatherDelay", "NASDelay").show(10)
df.printSchema()

```

###Trim & Standardize String Columns

```python


string_cols = ["Origin", "Dest", "UniqueCarrier", "TailNum", "CancellationCode"]

for c in string_cols:
    df = df.withColumn(c, trim(col(c)))

# Quick preview
df.select("Origin", "Dest", "UniqueCarrier", "TailNum").show(10, truncate=False)
```

### Remove Rows With Missing Critical Fields

```python
df = df.filter(
    col("Origin").isNotNull() &
    col("Dest").isNotNull() &
    col("UniqueCarrier").isNotNull()
)

print("Rows after critical field filtering:", df.count())

```

### Remove Impossible & Invalid Values

```python
df = df.filter(col("Distance") > 0)
df = df.filter(col("AirTime") > 0)

df = df.filter(col("Month").between(1, 12))
df = df.filter(col("DayOfWeek").between(1, 7))

# Remove unrealistic negative delays
df = df.filter(col("DepDelay") > -60)
df = df.filter(col("ArrDelay") > -60)

print("Rows after invalid value filtering:", df.count())
```

### Remove Duplicate Flight Records

```python
df = df.dropDuplicates([
    "Year",
    "Month",
    "DayofMonth",
    "FlightNum",
    "Origin",
    "Dest"
])

print("Rows after removing duplicates:", df.count())
```

## save results to the silver layer 

```python
# Save cleaned flights to silver layer
df.write.mode("overwrite").parquet(
    "abfss://lakehouse@airlinedata60104384.dfs.core.windows.net/processed/flights/"
)


silver_df = spark.read.parquet(
    "abfss://lakehouse@airlinedata60104384.dfs.core.windows.net/processed/flights/"
)
display(silver_df)
silver_df.printSchema()
silver_df.show(5)

```
